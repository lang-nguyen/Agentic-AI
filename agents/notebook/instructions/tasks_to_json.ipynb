```python
def task_to_dict(task, prompt_text: str):
    return {
        "user_id": task.user_id,
        "annotator": task.annotator,
        "actions": [
            {"name": action.name, "kwargs": action.kwargs} for action in task.actions
        ],
        "instruction": task.instruction,
        "outputs": task.outputs,
        "prompt": prompt_text.strip() if prompt_text else "",
    }
```

```python
import os

folder_path = "user_prompts_v2"

files = os.listdir(folder_path)
files_sorted = sorted(files, key=lambda x: int(x.split("_")[1].split(".")[0]))
print(files_sorted)
```

```python
prompts = []
for file in files_sorted:
    with open(os.path.join(folder_path, file), encoding="utf-8") as f:
        prompts.append(f.read())
print(len(prompts))
```

```python
from tasks_test import TASKS_TEST

assert len(TASKS_TEST) == len(prompts)

all_tasks = []
for task, prompt in zip(TASKS_TEST, prompts):
    task_dict = task_to_dict(task, prompt)
    all_tasks.append(task_dict)
print(len(all_tasks))
```

```python
import json

file_name = "tasks_test.json"
with open(file_name, "w", encoding="utf-8") as f:
    json.dump(all_tasks, f, indent=2)
```
