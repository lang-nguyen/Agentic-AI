```python
from pathlib import Path

from langchain_ollama import ChatOllama

from agentic_ai.pipeline.toolpolicy.tool_policy_generator import ToolPolicyGenerator
from agentic_ai.prompts import TAUBENCH_WIKI
from agentic_ai.tools.taubench_tools.cancel_pending_order import CancelPendingOrder
from agentic_ai.tools import ALL_TAUBENCH_TOOLS
```

```python
llm = ChatOllama(model="gpt-oss:20b", keep_alive=-1, num_ctx=8000)

tool = CancelPendingOrder.get_tool()

generator = ToolPolicyGenerator(
    llm=llm, policy_document=TAUBENCH_WIKI, tools=[tool], out_dir=Path(__name__).parent
)
```

```python
output_dir = Path().parent / ".local" / "output" / "short"

await ToolPolicyGenerator.run(
    policy_text=TAUBENCH_WIKI, tools=[tool], output_dir=output_dir, llm=llm, short=True
)
```

```python
output_dir = Path().parent / ".local" / "output" / "full"

await ToolPolicyGenerator.run(
    policy_text=TAUBENCH_WIKI, tools=[tool], output_dir=output_dir, llm=llm, short=False
)
```

```python
output_dir = Path().parent / ".local" / "output" / "all"

await ToolPolicyGenerator.run(
    policy_text=TAUBENCH_WIKI, tools=ALL_TAUBENCH_TOOLS, output_dir=output_dir, llm=llm, short=True
)
```
